# Data Cleaning: Removing Columns and Rows

In this section, we remove columns and rows with excessive missing values that are not essential for our analysis.

- **Columns to drop:** `license` (100% missing) and `house_rules` (50.81% missing)
- **Rows to drop:** Records with missing values in critical columns needed for analysis

```python
# Drop columns
df = df.drop(['license', 'house_rules'], axis=1)
```

```python
# Drop rows
columns_to_check = ['availability 365','calculated host listings count','review rate number', 'country', 'host name', 'minimum nights', 'host_identity_verified','service fee','price','NAME','Construction year','number of reviews','country code','instant_bookable','cancellation_policy','neighbourhood group','neighbourhood','long','lat']
df = df.dropna(subset=columns_to_check)
```

# Column Name Standardization

We standardize column names to ensure consistency and ease of use throughout the analysis:

1. **Strip whitespace** from column names
2. **Convert to lowercase** for uniformity
3. **Replace underscores** with spaces for readability

```python
df.columns
```

```python
df.columns = df.columns.str.strip()
```

```python
df.columns = df.columns.str.lower()
```

```python
df.columns = df.columns.str.replace('_', ' ')
```

# Handling Missing Values

We fill missing values in the dataset using appropriate imputation strategies:

- **`last review`**: Convert to datetime format and fill missing values with the median date
- **`reviews per month`**: Fill with median value after outlier analysis

```python
df['last review'] = pd.to_datetime(df['last review'])
df['last review'] = df['last review'].fillna(df['last review'].median())
```

```python
df.columns
```

## Outlier Detection and Imputation

The `reviews per month` column contains many outliers, as shown in the box plot below. We use the median for imputation to avoid skewing the data distribution, as the median is more robust to outliers than the mean.

```python
# Box Plot
plt.figure(figsize=(10, 4))
plt.boxplot(df['reviews per month'].dropna(), vert=False)
plt.xlabel('Reviews Per Month')
plt.title('Box Plot - Outlier Check')
plt.savefig('images/reviews_per_month_outlier_boxplot.png', dpi=300, bbox_inches='tight')
plt.close()
```

```python
df['reviews per month'] = df['reviews per month'].fillna(df['reviews per month'].median())
```

```python
df['service fee'].unique()
```

# Data Type Conversion: Cleaning Price Fields

We clean price-related columns to enable mathematical operations:

1. Remove currency symbols (`$`)
2. Remove whitespace
3. Remove thousand separators (`,`)
4. Convert to numeric data types (`int64` for service fee, `float` for price)

This transformation allows us to perform statistical analysis and calculations on pricing data.

```python
df['service fee'] = df['service fee'].str.replace('$', '')
df['service fee'] = df['service fee'].str.replace(' ', '')

df['service fee'] = df['service fee'].astype('int64')
```

```python
df['price']
```

```python
df['price'].unique().tolist()
```

```python
df['price'].value_counts()
```

```python
df['price'] = df['price'].str.replace('$', '')
df['price'] = df['price'].str.replace(' ', '')
df['price'] = df['price'].str.replace(',', '')

df['price'] = df['price'].astype(float)
```

```python
df['price'].unique().tolist()
```

```python
df['instant bookable'].unique()
df['instant bookable'] = df['instant bookable'].astype(bool)
```

```python
df['neighbourhood group'].unique().tolist()
```

```python
print(df['neighbourhood group'].unique())
df['neighbourhood group']=df['neighbourhood group'].str.replace('brookln', 'Brooklyn')
df['neighbourhood group'].value_counts()
```

# Removing Duplicate Records

We identify and remove duplicate rows from the dataset to ensure data quality and prevent redundant information from affecting our analysis. This step helps maintain the integrity of our statistical calculations.

```python
df = df.drop_duplicates()
```

```python
df.shape
```

# Identifying Outliers and Invalid Data

We examine the distribution of key variables to identify:

- **Outliers**: Extreme values that may skew our analysis
- **Invalid data**: Impossible or incorrect values that need to be addressed
- **Data quality issues**: Inconsistencies that require cleaning

This analysis helps us understand the data quality and determine appropriate handling strategies.

```python
df['availability 365'].unique().tolist()
```

```python
df.loc[(df["availability 365"] < 0) | (df["availability 365"] > 365), "availability 365"] = np.nan
df = df.dropna(subset=['availability 365'])
```

```python
df['minimum nights'].unique().tolist()
```

```python
df.loc[(df["minimum nights"] < 1) | (df["minimum nights"] > 365), "minimum nights"] = np.nan
df = df.dropna(subset=['minimum nights'])
```

```python
df.isna().sum()
```

```python
info = pd.DataFrame({
    "col": df.columns,
    "dtype": df.dtypes.astype(str).values,
    "missing": df.isna().sum().values
})
info["missing_%"] = (info["missing"]/len(df)*100).round(2)
info.sort_values("missing_%", ascending=False)
```

# Data Visualization

In this section, we create visualizations to explore patterns and relationships in the Airbnb dataset.

## 📊 CHART 1 — Price Distribution

**Objective:** Visualize the distribution of prices and identify outliers in the dataset.

```python
plt.figure(figsize=(9,5))

sns.set_style("whitegrid")
sns.histplot(df['price'], kde=True, bins=20, )
plt.xlabel("Price")
plt.ylabel("Count")
plt.title("Price Distribution")

plt.savefig('images/price_distribution_histogram.png', dpi=300, bbox_inches='tight')
plt.close()
```

## 📊 CHART 2 — Price by Room Type

**Objective:** Examine the impact of room type on pricing to understand how different accommodation types are priced.

```python
price_room = df.groupby("room type")["price"].median().sort_values(ascending=False)

plt.figure(figsize=(7,4))
plt.bar(price_room.index, price_room.values)
plt.xlabel("Room Type")
plt.ylabel("Median Price")
plt.title("Median Price by Room Type")
plt.grid(True, axis="y")
plt.savefig('images/median_price_by_room_type.png', dpi=300, bbox_inches='tight')
plt.close()
```

## 📊 CHART 3 — Room Type Distribution

**Objective:** Understand which room types are most common in the dataset and their relative proportions.

```python
room_counts = df["room type"].value_counts().sort_values(ascending=False)

display(room_counts)
plt.figure(figsize=(8, 5))
plt.bar(room_counts.index, room_counts.values)
plt.xlabel("Room Type")
plt.ylabel("Count")
plt.title("Distribution of Room Types")
plt.xticks(rotation=45, ha='right')
plt.grid(True, axis="y", alpha=0.3)

plt.tight_layout()
plt.savefig('images/room_type_distribution.png', dpi=300, bbox_inches='tight')
plt.close()
```

## 📊 CHART 4 — Price by Neighbourhood Group

**Objective:** Analyze the impact of location (neighbourhood group) on pricing across different areas of the city.

```python
price_ng = df.groupby("neighbourhood group")["price"].median().sort_values(ascending=False)

plt.figure(figsize=(8,4))
plt.bar(price_ng.index, price_ng.values)
plt.xticks(rotation=30, ha="right")
plt.xlabel("Neighbourhood Group")
plt.ylabel("Median Price")
plt.title("Median Price by Neighbourhood Group")

plt.grid(True)
plt.savefig('images/median_price_by_neighbourhood.png', dpi=300, bbox_inches='tight')
plt.close()
```

## 📊 CHART 5 — Instant Bookable vs Price

**Objective:** Determine whether instant booking availability creates a price difference and affects listing pricing strategy.

```python
ib_price = df.groupby("instant bookable")["price"].median()

plt.figure(figsize=(6,4))
plt.bar(ib_price.index.astype(str), ib_price.values)
plt.xlabel("Instant Bookable")
plt.ylabel("Median Price")
plt.title("Price vs Instant Bookable")
plt.grid(True, axis="y")
plt.savefig('images/price_vs_instant_bookable.png', dpi=300, bbox_inches='tight')
plt.close()
```

## 📊 CHART 6 — Reviews per Month (Demand Indicator)

**Objective:** Identify which neighbourhoods are more active by analyzing review frequency as a proxy for demand.

```python
rpm_ng = df.groupby("neighbourhood group")["reviews per month"].median().sort_values(ascending=False)

plt.figure(figsize=(8,4))
plt.bar(rpm_ng.index, rpm_ng.values)
plt.xticks(rotation=30, ha="right")
plt.xlabel("Neighbourhood Group")
plt.ylabel("Median Reviews per Month")
plt.title("Demand by Neighbourhood Group")
plt.grid(True, axis="y")
plt.savefig('images/demand_by_neighbourhood.png', dpi=300, bbox_inches='tight')
plt.close()
```

## 📊 CHART 7 — Multi-Host Effect

**Objective:** Examine the pricing strategy of hosts with multiple listings to understand their market approach.

```python
host_price = df.groupby("calculated host listings count")["price"].median().head(10).sort_values(ascending=False)

plt.figure(figsize=(7,4))
plt.bar(host_price.index, host_price.values)
plt.xlabel("Host Listing Count")
plt.ylabel("Median Price")
plt.title("Multi-host Pricing Behavior")
plt.grid(True, axis="y")
plt.savefig('images/multihost_pricing_behavior.png', dpi=300, bbox_inches='tight')
plt.close()
```

# Exploratory Data Analysis Summary

This study analyzed the Airbnb Open Data to examine pricing, demand, and market structure. The analysis revealed that **room type** and **location** have the most significant impact on pricing.

## Key Findings:

- **Instant Bookable Listings:** Tend to be priced higher, suggesting a premium for convenience
- **Availability:** Listings with lower availability tend to be more expensive, indicating higher demand
- **Central Locations:** Show both higher prices and higher demand (more reviews)
- **Multi-Host Strategy:** Users with multiple listings employ more competitive pricing strategies

## Implications:

These findings provide a strong foundation for:
- Building price prediction models
- Conducting market analysis
- Understanding competitive dynamics in the short-term rental market

The insights can help both hosts optimize their pricing strategies and guests make informed booking decisions.
